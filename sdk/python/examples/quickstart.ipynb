# Data Connect Hub quickstart

Requires a running DCH stack.

Optional env overrides: `DCH_REST_URL`, `DCH_FLIGHT_URL`, `DCH_TOKEN`, `DCH_TENANT_ID`.

Use the `sdk/python/.venv` kernel.

```python
import os

from data_connect_hub import DataConnectClient

REST_URL = os.getenv("DCH_REST_URL", "http://localhost:8080")
FLIGHT_URL = os.getenv("DCH_FLIGHT_URL", "grpc://localhost:50051")
TOKEN = os.getenv("DCH_TOKEN", "")
TENANT_ID = os.getenv("DCH_TENANT_ID", "default")

client = DataConnectClient(
    rest_url=REST_URL,
    flight_url=FLIGHT_URL,
    token=TOKEN,
    tenant_id=TENANT_ID,
)
```

## Connection types (REST)

```python
types = client.list_connection_types()
print(f"Found {len(types)} connection type(s):")
for ct in types:
    print(f"  [{ct.id}] {ct.name}: {ct.description}")
```

```python
new_type = client.create_connection_type(
    name="example-postgres-3",
    provider="postgres",
    description="PostgreSQL connector created by quickstart example",
)
print(f"Created type: {new_type.id} ({new_type.name})")
```

```python
fetched = client.get_connection_type(new_type.id)
print(f"Fetched: {fetched.name} — {fetched.description}")

types = client.list_connection_types()
print(f"There are {len(types)} connection type(s)")
```

## Flight SQL

```python
info = client.server_info()
print("Server info:")
for key, value in info.items():
    print(f"  {key}: {value}")
```
